## Un calcul, une fonction et une vérification

Le flux lumineux fourni est $S=L/(4\pi d^2)$.

```python
def flux(L,d):
    return L/(4*pi*d*d)

L=3.828e26;d=1.496e11
S=flux(L,d)
print("Flux reçu :",round(S,1),"W/m²")
assert isclose(flux(L,2*d),S/4)
print("La distance doublée divise le flux par quatre.")
```

```python
distances=[j*.1*d for j in range(5,21)]
plt.plot([x/d for x in distances],[flux(L,x) for x in distances])
plt.xlabel("Distance / distance de référence")
plt.ylabel("Flux reçu (W/m²)")
plt.grid();plt.show()
```
